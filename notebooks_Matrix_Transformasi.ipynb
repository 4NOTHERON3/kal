# Transformasi Linear dan Translasi

## 1. Koordinat Titik

$$

$$

$$
D = (2,4), \quad E = (2,0), \quad F = (4,0)
$$

## 2. Bentuk Umum Transformasi

Misalkan transformasi linear:

$$
T(x,y) = (ax + by,\; cx + dy)
$$

## 3. Menyusun Sistem Persamaan

### Dari A → B

$$
T(2,3) = (2,1)
$$

$$
2a + 3b = 2
$$

$$
2c + 3d = 1
$$

### Dari D → E

$$
T(2,4) = (2,0)
$$

$$
2a + 4b = 2
$$

$$
2c + 4d = 0
$$

## 4. Menyelesaikan Sistem

### Untuk a dan b

$$
(2a + 4b) - (2a + 3b) = 2 - 2
$$

$$
b = 0
$$

$$
2a = 2 \Rightarrow a = 1
$$

### Untuk c dan d

$$
(2c + 4d) - (2c + 3d) = 0 - 1
$$

$$
d = -1
$$

$$
2c + 3(-1) = 1
$$

$$
2c - 3 = 1 \Rightarrow 2c = 4 \Rightarrow c = 2
$$

## 7. Pengujian Transformasi

### A → B

$$
T(2,3) = (2,\;4-3) = (2,1)
$$

 sesuai

### B → C

$$

$$

### D → E

$$

 sesuai

### E → F

$$

$$

$$
(2,4) \neq (4,0)
$$

# Transformasi dan Titik Bayangan

## Diketahui Transformasi

Transformasi utama:

$$
T_1(x,y) = (x,\; 2x - y)
$$

Translasi:

$$
T_2(x,y) = (x+2,\; y)
$$

## 1. Hitung Titik Bayangan (menggunakan $T_1$)

### A'

$$
A = (2,3)
$$

$$
A' = T_1(2,3) = (2,\; 2(2) - 3) = (2,\; 4 - 3) = (2,1)
$$

### B'

$$
B = (2,1)
$$

$$
B' = T_1(2,1) = (2,\; 2(2) - 1) = (2,\; 4 - 1) = (2,3)
$$

### C'

$$
C = (4,1)
$$

$$
C' = T_1(4,1) = (4,\; 2(4) - 1) = (4,\; 8 - 1) = (4,7)
$$

### D'

$$
D = (2,4)
$$

$$
D' = T_1(2,4) = (2,\; 2(2) - 4) = (2,\; 4 - 4) = (2,0)
$$

### E'

$$
E = (2,0)
$$

$$
E' = T_1(2,0) = (2,\; 2(2) - 0) = (2,\; 4 - 0) = (2,4)
$$

### F'

$$
F = (4,0)
$$

$$
F' = T_1(4,0) = (4,\; 2(4) - 0) = (4,\; 8 - 0) = (4,8)
$$
